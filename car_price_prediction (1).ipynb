```python
# libraries importing 
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,OrdinalEncoder # for handling categorical data
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error,root_mean_squared_error # for model evaluation
```

```python
# data loading 
df=pd.read_csv(r"C:\Users\INDIAN\Downloads\AI_projects\car_price_predictor\data\cardekho.csv")
df.head()
```

```python
# drop un wanted columns

df.drop(columns=["name","max_power"],inplace=True)
```

```python
df.head()
```

```python
# rename mileage(km/ltr/kg) to mileage
df.rename(columns={"mileage(km/ltr/kg)":"mileage"},inplace=True)
```

```python
# filling null values ''
df["mileage"]=df["mileage"].fillna(df["mileage"].mean())
df["engine"]=df["engine"].fillna(df["engine"].mean())
df["seats"]=df["seats"].fillna(df["seats"].mean())
```

```python
# encoding categorical columns
label=LabelEncoder()  # initialize label encoder 

df["enoded_fuel"]=label.fit_transform(df["fuel"])
var=dict(zip(df["fuel"],df["enoded_fuel"]))
print(var)

# encoding seler type
df["encoded_seller"]=label.fit_transform(df["seller_type"])
var1=dict(zip(df["seller_type"],df["encoded_seller"]))
print(var1)

# encoding transmission ciolumn
df["encoded_trans"]=label.fit_transform(df["transmission"])
var2=dict(zip(df["transmission"],df["encoded_trans"]))
print(var2)

# encodeing owner column
df["encoded_owner"]=label.fit_transform(df["owner"])
var3=dict(zip(df["owner"],df["encoded_owner"]))
print(var3)
```

```python
df.isnull().sum()
```

```python
# feature selection
x=df.drop(columns=["selling_price","fuel","seller_type","transmission","owner"])
y=df["selling_price"]
```

```python
x.isnull().sum(),y.isnull().sum()
```

```python
# splitting data 
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model selection 
model=LinearRegression()

# model training
model.fit(x_train,y_train)
```

```python
# test model on test data
y_pred=model.predict(x_test)
r2_score(y_test,y_pred)
```

```python
# accuracy on train data 
y_pred2=model.predict(x_train)
r2_score(y_train,y_pred2)
```
